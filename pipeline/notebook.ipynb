```python
prefix = 'https://github.com/DataTalksClub/nyc-tlc-data/releases/download/yellow/'
data_url = f'{prefix}/yellow_tripdata_2021-01.csv.gz'

print(data_url)
```

```python
df = pd.read_csv(data_url)
df.head()
```

```python
dtype = {
    "VendorID": "Int64",
    "passenger_count": "Int64",
    "trip_distance": "float64",
    "RatecodeID": "Int64",
    "store_and_fwd_flag": "string",
    "PULocationID": "Int64",
    "DOLocationID": "Int64",
    "payment_type": "Int64",
    "fare_amount": "float64",
    "extra": "float64",
    "mta_tax": "float64",
    "tip_amount": "float64",
    "tolls_amount": "float64",
    "improvement_surcharge": "float64",
    "total_amount": "float64",
    "congestion_surcharge": "float64"
}

parse_dates = [
    "tpep_pickup_datetime",
    "tpep_dropoff_datetime"
]

df = pd.read_csv(
    data_url,
    dtype=dtype,
    parse_dates=parse_dates
)

df.head()
```

```python
print(df.dtypes)
print(df.shape)
```

```python
from sqlalchemy import create_engine

engine = create_engine('postgresql://root:root@localhost:5432/ny_taxi')
```

```python
print(pd.io.sql.get_schema(df, name="yellow_taxi_data", con = engine))
```

```python
df.head(0)
```

```python
df.head(0).to_sql(name = 'yellow_taxi_data', con = engine, if_exists= 'replace')
```

```python
df_iter = pd.read_csv( data_url,
                       dtype = dtype,
                       parse_dates = parse_dates,
                       iterator = True,
                       chunksize = 100000,
                     )
```

```python
df_iter
```

```python
for df_chunk in df_iter:
    print(len(df_chunk))
```

```python
! uv add tqdm
```

```python
from tqdm.auto import tqdm
```

```python
for df_insert in df_iter:
    df_insert.to_sql(name='yellow_taxi_data',con= engine, if_exists='append')
```

```python
first_chunk = next(df_iter)

first_chunk.head(0).to_sql(
    name="yellow_taxi_data",
    con=engine,
    if_exists="replace"
)
print(f"Table is created!")

first_chunk.to_sql(
    name="yellow_taxi_data",
    con=engine,
    if_exists="append"
)
print(f"Inserted rows:{len(first_chunk)}")

for df_chunks in tqdm(df_iter):
    df_chunks.to_sql(
        name="yellow_taxi_data",
        con=engine,
        if_exists="append"
    )
    print(f"Inserted rows:{len(df_chunks)}")
```
